```python
from google.colab import drive
import os

drive.mount('/content/drive')

BASE_PATH = '/content/drive/MyDrive'
BASE_PATH2 = '/content/drive/MyDrive/Commit & Run'

IMAGE_DIR = os.path.join(BASE_PATH2, 'images')
LABEL_FILE = os.path.join(BASE_PATH, 'labels.txt')
```

```python
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import models, transforms
from torchvision.models import ResNet18_Weights
from PIL import Image
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1. Custom Dataset Class
class MultilabelDataset(Dataset):
    def __init__(self, txt_file, img_dir, transform=None):
        self.img_dir = img_dir
        self.transform = transform

        # Read the labels file
        with open(txt_file, 'r') as f:
            lines = f.readlines()

        self.data = []
        missing_count = 0

        for line in lines:
            parts = line.strip().split()
            if len(parts) >= 5:
                img_name = parts[0]
                img_path = os.path.join(self.img_dir, img_name)

                if os.path.exists(img_path):
                    # Convert "NA" to -1.0, otherwise to float
                    labels = [float(x) if x != 'NA' else -1.0 for x in parts[1:5]]
                    self.data.append((img_name, labels))
                else:
                    missing_count += 1

        if missing_count > 0:
            print(f"Note: Skipped {missing_count} images that were in labels.txt but missing from the folder.")

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        img_name, labels = self.data[idx]
        img_path = os.path.join(self.img_dir, img_name)

        image = Image.open(img_path).convert('RGB')

        if self.transform:
            image = self.transform(image)

        return image, torch.tensor(labels, dtype=torch.float32)

# 2. Handling Imbalance & Transforms
# Transforms for fine-tuning on ImageNet weights [cite: 18]
train_transforms = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.RandomHorizontalFlip(),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
])

dataset = MultilabelDataset(LABEL_FILE, IMAGE_DIR, transform=train_transforms)
dataloader = DataLoader(dataset, batch_size=16, shuffle=True)


all_labels = np.array([labels for _, labels in dataset.data])
pos_weights = []
for i in range(4):
    valid_mask = all_labels[:, i] != -1.0
    valid_labels = all_labels[:, i][valid_mask]
    num_pos = (valid_labels == 1.0).sum()
    num_neg = (valid_labels == 0.0).sum()
    weight = num_neg / num_pos if num_pos > 0 else 1.0
    pos_weights.append(weight)

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
pos_weights_tensor = torch.tensor(pos_weights, dtype=torch.float32).to(device)

# 3. Model Architecture setup

model = models.resnet18(weights=ResNet18_Weights.DEFAULT)
num_ftrs = model.fc.in_features

model.fc = nn.Linear(num_ftrs, 4)
model = model.to(device)

# Using BCEWithLogitsLoss
criterion = nn.BCEWithLogitsLoss(reduction='none', pos_weight=pos_weights_tensor)
optimizer = optim.Adam(model.parameters(), lr=0.001)

# 4. Training Loop
epochs = 10
iteration_losses = []

print("Starting training...")
model.train()
iteration_count = 0

for epoch in range(epochs):
    for images, labels in dataloader:
        images, labels = images.to(device), labels.to(device)

        optimizer.zero_grad()
        outputs = model(images)

        raw_loss = criterion(outputs, labels)


        valid_mask = (labels != -1.0).float()

        masked_loss = raw_loss * valid_mask
        loss = masked_loss.sum() / valid_mask.sum().clamp(min=1e-6)

        loss.backward()
        optimizer.step()

        iteration_losses.append(loss.item())
        iteration_count += 1

    print(f"Epoch {epoch+1}/{epochs} completed.")

torch.save(model.state_dict(), 'multilabel_model_weights.pth')
print("Model weights saved to multilabel_model_weights.pth")

# 5. Loss Curve Plot

plt.figure(figsize=(10, 5))
plt.plot(range(1, iteration_count + 1), iteration_losses)

plt.xlabel("iteration_number")
plt.ylabel("training_loss")
plt.title("Aimonk_multilabel_problem")
plt.grid(True)
plt.savefig('loss_curve.png')
plt.show()
```

```python
# Inference
def predict_attributes(image_path, model_path):
    device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

    model = models.resnet18(weights=None)
    model.fc = nn.Linear(model.fc.in_features, 4)
    model.load_state_dict(torch.load(model_path, map_location=device))
    model = model.to(device)
    model.eval()

    val_transforms = transforms.Compose([
        transforms.Resize((224, 224)),
        transforms.ToTensor(),
        transforms.Normalize(mean=[0.485, 0.456, 0.406], std=[0.229, 0.224, 0.225])
    ])

    # Load and process image
    image = Image.open(image_path).convert('RGB')
    image_tensor = val_transforms(image).unsqueeze(0).to(device)

    # Predict
    with torch.no_grad():
        outputs = model(image_tensor)

        probs = torch.sigmoid(outputs).squeeze().cpu().numpy()

    present_attributes = []
    for i, prob in enumerate(probs):
        if prob > 0.5:
            present_attributes.append(f"Attr{i+1}")

    print(f"Attributes present in {os.path.basename(image_path)}: {present_attributes}")

```

```python
test_image_path = os.path.join(IMAGE_DIR, 'image_899.jpg')
model_weights_path = 'multilabel_model_weights.pth'

predict_attributes(test_image_path, model_weights_path)
```
